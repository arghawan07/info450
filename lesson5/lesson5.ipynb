```python
import requests

url = "https://raw.githubusercontent.com/arghawan07/info450/refs/heads/main/lesson5/worldcup.json"
response = requests.get(url)

print(response.status_code)  # should print 200

data = response.json()  # this should work now without errors
print(type(data))
print(data.keys())

print(len(data["matches"]), "matches loaded")

last_match = data["matches"][-1]
print(last_match)
```

```python
# Loop through the list of matches in the JSON
for match in data["matches"]:
    team1 = match["team1"]
    team2 = match["team2"]
    score = match["score"]["ft"] # Full-time score format: [team1_goals, team2_goals]

    print(f"{team1} vs {team2} | Final Score: {score}")
```

```python
# Check top-level type
type(data)

# Check dictionary keys
print("Keys:", data.keys())

# Extract matches list
matches = data["matches"]
print("Matches Type:", type(matches))

# Get the last match using [-1]
last_match = matches[-1]
print("Last Match:", last_match)

# Check last match keys and specifics
print("Last Match Keys:", last_match.keys())
print("Teams:", last_match["team1"], "vs", last_match["team2"])
print("Final Score:", last_match["score"]["ft"])
```
